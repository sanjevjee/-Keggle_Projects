```python
# importing necessary library
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# loading the dataset
# The dataset is a CSV file containing tweets labelled with sentiment
data = pd.read_csv('/Users/mac/Downloads/data.csv',encoding='ISO-8859-1')
df = data.copy()
```

```python
# Displaying the first few rows of the dataset
df.head()
```

```python
df.info()
```

```python
train=df[df["Date"] < "20150101"]
test=df[df["Date"] >= "20141231"]
```

```python
# doing some feature engineering
data=train.iloc[:,2:27]
data.replace("[^a-zA-Z]", '', regex=True, inplace=True)
# renaming the columns
list1=[i for i in range(25)] 
new_index=[str(i) for i in list1]
data.columns=new_index
data.head()
```

```python
# converting headlines to lower case
for index in data.columns:
    data[index]=data[index].str.lower()
data.head()
```

```python
' '.join(str(x ) for x in data.iloc[1,0:25])
```

```python
headlines=[]
for row in range(0, len(data.index)):
    headlines.append(' '.join(str(x) for x in data.iloc[row,0:25]))
```

```python
headlines[1]
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.ensemble import RandomForestClassifier
```

```python
countvector=CountVectorizer(ngram_range=(2,2))
traindataset=countvector.fit_transform(headlines)
```

```python
randomclassifier=RandomForestClassifier(n_estimators=200, random_state=0,criterion='entropy')
randomclassifier.fit(traindataset,train['Label'])
```

```python
# predict the test set
testdata=test.iloc[:,2:27]
testdata.replace("[^a-zA-Z]", '', regex=True, inplace=True)
testdata.columns=new_index  
for index in testdata.columns:
    testdata[index]=testdata[index].str.lower()
testheadlines=[]
for row in range(0, len(testdata.index)):
    testheadlines.append(' '.join(str(x) for x in testdata.iloc[row,0:25]))
testheadlines[1]
testdataset=countvector.transform(testheadlines)
predictions=randomclassifier.predict(testdataset)
predictions[1]
```

```python
import sklearn
from sklearn.metrics import confusion_matrix
sklearn.metrics.confusion_matrix(test['Label'], predictions)
from sklearn.metrics import classification_report
print(classification_report(test['Label'], predictions))
from sklearn.metrics import accuracy_score
accuracy=accuracy_score(test['Label'], predictions)
print(accuracy)
```
